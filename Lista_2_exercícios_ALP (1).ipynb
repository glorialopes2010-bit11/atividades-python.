Questão 04 - Estacionamento

```python
horas = int(input("Digite as horas:"))
if horas <= 2:
 valor = 8
elif horas <= 5:
 valor = 8 + 3 * (horas - 2)
else:
 valor = 17 + 2 * (horas - 5)

print("valor a pagar R$" , valor)
```

Questão 05 - Máquina de venda de ingresso

```python
preço = 25
total = 0

while total < preço:
  valor = int(input("Insira um valor:"))
  total += valor
print("total inserido:", total)
print(" troco: R$",total - preço)
```

Questão 08 - Detectando repetição consecutiva

```python
anterior = int(input("número: "))
tem_iguais = False

for i in range(7):
  atual = int(input("Número: "))
  if atual == anterior:
    tem_iguais = True
  anterior = atual

if tem_iguais:
  print("Há números iguais e consecutivos")
else:
  print("Não há números iguais e consecutivos")

```

Questão 09 - Controle de saldo

```python
saldo = 500.0
for i in range(6):
    tipo = int(input("Tipo (1 = depósito, 2 = saque)"))
    valor = float(input("Valor: "))
    if tipo == 1:
     saldo += valor
    elif tipo == 2:
     if valor <= saldo:
      saldo -= valor
     else:
       print("Saldo insuficiente")
print("Saldo final:", saldo)
```

Questão 11 - Número primo

```python
n = int(input("Digite um número maior que 1: "))
primo = True
i = 2

while i < n:
  if n % i == 0:
    primo = False
  i += 1
if primo:
  print("É primo")
else:
  print("Não é primo")
```

Questão 12 - Divisores de um número

```python
n = int(input("Numero: "))
quantidade = 0

for i in range(1, n + 1):
  if n % i == 0:
    print(i)
    quantidade += 1
print("Quantidade de divisores:", quantidade)
```

Questão 13 - Números primos em um intervalo

```python
a = int(input("Numero: "))
b = int(input("Numero: "))
quantidade = 0

for num in range(a, b + 1):
  if num > 1:
    primo = True
    i = 2
    while i < num:
      if num % i == 0:
        primo = False
      i += 1
    if primo:
      print(num)
      quantidade += 1
print("Quantidade de primos: ", quantidade)
```

Questão 14 - Jogo com limite de tentativas

```python

n_secreto = 37
acertou = False

for i in range(1,6):
 palpite = int(input("Palpite " + str(i) +": "))
 if palpite == n_secreto:
  print("Acertou!")
  break
  acertou = True

 elif palpite < n_secreto:
  print("Tente um palpite maior: ")

 else:
  print("Tente um palpite menor: ")

if not acertou:
  print("Fim das tentativas")
```

Questão 15 - Desafio final - Caixa de atendimento


```python
saldo = 200.0
entrada = 0
retiradas_recusadas = 0
retiradas_aceitas = 0
total_entrou = 0
total_saiu = 0

código = int(input("Digite código: 1 = entrada, 2 = retirada, 0 = sair :"))

while código != 0:
  if código == 1:
    valor = float(input("Digite o valor: "))
    saldo += valor
    entrada += 1
    total_entrou += valor

  elif código == 2:
    valor = float(input("Digite o valor: "))
    if valor <= saldo:
      saldo -= valor
      retiradas_aceitas += 1
      total_saiu += valor

    else:
      print("Saldo insuficiente")
      retiradas_recusadas += 1
  código = int(input("Digite código: 1 = entrada, 2 = retirada, 0 = sair :"))

print("Saldo: ", saldo)
print("Entradas: ", entrada)
print("Retiradas recusadas: ", retiradas_recusadas)
print("Retiradas aceitas: ", retiradas_aceitas)
print("Total entrado: ", total_entrou)
print("Total retirado: ", total_saiu)
```
