```python
#5 - Situação do aluno
nota = int(input("Primeira nota do aluno:"))
if nota >= 7:
  print("Está Aprovado!")
elif 5 <= nota < 7:
  print("Está de recuperação")
elif nota < 5:
  print("Está reprovado")

nota2 = int(input("Segunda nota do aluno:"))
if nota2 >= 7:
  print("Está aprovado!")
elif 5 <= nota2 < 7:
  print("Está de recuperação")
elif nota < 5:
  print("Está reprovado")
#média calculada
média = (nota + nota2) / 2
print("A média do aluno é:",média)
```

```python
#7 - Contagem crescente
n = int(input("Digite um número:"))
for n in range(1,n + 1):
  print(n)
```

```python
#8 - Somando de 1 até n
n = int(input("Digite um número:"))
soma = 0
for number in range(1, n + 1):
  soma += number
print(soma)
```

```python
#11 - Maior valor digitado
maior = int(input("Digite um número:"))
quantidade = 2
while quantidade < 6:
  num = int(input("Digite outro número:"))

  if num > maior:
    maior = num
  quantidade += 1
print("o maior é:",maior)
```

```python
#14 - Número secreto
secreto = 7
numero = int(input("Digite e descubra o número:"))
while numero != secreto:
  if numero < secreto:
    print("tente um número maior!:")
  elif numero > secreto:
    print("tente um número menor!:")
  numero = int(input("tente novamente!:"))

print("Parabéns você acertou")
```
